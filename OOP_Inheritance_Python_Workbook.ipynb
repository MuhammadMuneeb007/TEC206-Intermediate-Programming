# 1. What is a Programming Paradigm?

A **programming paradigm** is a general style or way of organising programs.

Different paradigms encourage programmers to think about problems differently.

In practice, Python is a **multi-paradigm language**. A Python program can combine multiple styles.

## 1.1 Functional Programming

Functional programming focuses on:

- Functions as first-class values
- Transforming data through functions
- Avoiding unnecessary changes to shared state
- Expressions such as `map`, `filter`, comprehensions, and higher-order functions

### Simple example

We want to square a collection of numbers.

````python
numbers = [1, 2, 3, 4, 5]

squared = list(map(lambda x: x ** 2, numbers))

print("Original:", numbers)
print("Squared :", squared)
````

The same task can also be written using a list comprehension:

````python
numbers = [1, 2, 3, 4, 5]
squared = [x ** 2 for x in numbers]

print(squared)
````

## 1.2 Logic Programming

Logic programming describes **facts**, **rules**, and relationships, then asks what conclusions follow.

Languages such as **Prolog** are strongly associated with logic programming.

Python is not primarily a logic-programming language, but we can demonstrate the basic idea with facts and rules.

````python
# Facts
parent_of = {
    "Ali": ["Sara", "Omar"],
    "Sara": ["Noah"]
}

def is_parent(parent, child):
    return child in parent_of.get(parent, [])

def is_grandparent(grandparent, grandchild):
    for child in parent_of.get(grandparent, []):
        if grandchild in parent_of.get(child, []):
            return True
    return False

print("Is Ali a parent of Sara?", is_parent("Ali", "Sara"))
print("Is Ali a grandparent of Noah?", is_grandparent("Ali", "Noah"))
````

## 1.3 Event-Driven Programming

Event-driven programs respond to **events**.

Examples of events include:

- A button click
- A keyboard press
- A mouse movement
- A network request
- A message arriving
- A timer finishing

Graphical applications, games, web applications, and mobile applications commonly use event-driven programming.

````python
def on_button_click():
    print("Button clicked! Running the event handler...")

# Simulating an event:
on_button_click()
````

## 1.4 Object-Oriented Programming

Object-oriented programming (OOP) organises software around **objects**.

An object usually contains:

- **Data / state** → attributes
- **Behaviour** → methods

A **class** is a blueprint.  
An **object** is an instance created from that blueprint.

````python
class Student:
    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id

    def introduce(self):
        return f"My name is {self.name} and my student ID is {self.student_id}."


student_1 = Student("Aisha", "S1001")
student_2 = Student("Daniel", "S1002")

print(student_1.introduce())
print(student_2.introduce())
````

## 1.5 Quick Comparison of Paradigms

| Paradigm | Main idea | Typical question |
|---|---|---|
| Functional | Transform data using functions | "What transformation should be applied?" |
| Logic | State facts/rules and infer conclusions | "What is true given these rules?" |
| Event-driven | Respond to events | "What should happen when this event occurs?" |
| Object-oriented | Model data and behaviour as objects | "What objects exist and how do they interact?" |

These paradigms are not mutually exclusive. A real Python application may use all of them.

# 2. OOP Foundations: Class, Object, Attribute, and Method

Before discussing inheritance, we need four basic terms.

### Class
A blueprint that defines common data and behaviour.

### Object
A concrete instance of a class.

### Attribute
A value stored in an object.

### Method
A function defined inside a class.

````python
class Car:
    # Constructor
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    # Method
    def description(self):
        return f"{self.year} {self.brand} {self.model}"


my_car = Car("Toyota", "Corolla", 2024)

print("Brand:", my_car.brand)
print("Model:", my_car.model)
print("Description:", my_car.description())
````

# 3. The Four Pillars of Object-Oriented Programming

The four commonly discussed OOP pillars are:

1. **Encapsulation**
2. **Abstraction**
3. **Inheritance**
4. **Polymorphism**

This lecture will briefly introduce all four, then focus mainly on **inheritance**.

## 3.1 Encapsulation

**Encapsulation** means grouping data and the methods that operate on that data inside a class.

It can also involve controlling how internal state is accessed or modified.

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance  # underscore = intended as internal/protected by convention

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be greater than zero.")
        self._balance += amount

    def get_balance(self):
        return self._balance


account = BankAccount("Maya", 100)
account.deposit(50)

print(account.get_balance())
````

## 3.2 Abstraction

**Abstraction** means showing the essential interface while hiding unnecessary implementation details.

A user of a class should know **what** a method does without needing to know every internal step.

````python
from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class CreditCardPayment(PaymentMethod):
    def pay(self, amount):
        return f"Paid ${amount:.2f} using a credit card."


payment = CreditCardPayment()
print(payment.pay(49.95))
````

## 3.3 Polymorphism

**Polymorphism** means "many forms."

Different objects can respond to the same method name in different ways.

````python
class Dog:
    def speak(self):
        return "Woof!"


class Cat:
    def speak(self):
        return "Meow!"


animals = [Dog(), Cat()]

for animal in animals:
    print(animal.speak())
````

## 3.4 Inheritance

**Inheritance** allows one class to reuse and extend the attributes and methods of another class.

Terminology:

- **Parent class / base class / superclass** → the class being inherited from
- **Child class / derived class / subclass** → the class that inherits

Inheritance models an **"is-a"** relationship.

Examples:

- A `Dog` **is an** `Animal`.
- A `Car` **is a** `Vehicle`.
- A `Manager` **is an** `Employee`.

# 4. Inheritance: Core Idea

Consider a system containing different kinds of animals.

Without inheritance, we might duplicate common code:

````python
class DogWithoutInheritance:
    def __init__(self, name):
        self.name = name

    def eat(self):
        return f"{self.name} is eating."

    def bark(self):
        return f"{self.name} says woof!"


class CatWithoutInheritance:
    def __init__(self, name):
        self.name = name

    def eat(self):
        return f"{self.name} is eating."

    def meow(self):
        return f"{self.name} says meow!"
````

Notice that both classes repeat:

- `name`
- the constructor
- `eat()`

Inheritance lets us move shared behaviour into one parent class.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        return f"{self.name} is eating."


class Dog(Animal):
    def bark(self):
        return f"{self.name} says woof!"


class Cat(Animal):
    def meow(self):
        return f"{self.name} says meow!"


dog = Dog("Buddy")
cat = Cat("Luna")

print(dog.eat())   # inherited from Animal
print(dog.bark())  # defined in Dog

print(cat.eat())   # inherited from Animal
print(cat.meow())  # defined in Cat
````

## 4.1 What exactly did the child class inherit?

`Dog` and `Cat` inherit the accessible behaviour defined in `Animal`.

Because `Dog` does not define its own `__init__`, Python uses the constructor inherited from `Animal`.

That is why this works:

```python
dog = Dog("Buddy")
```

even though the `Dog` class itself contains no constructor.

````python
print("Dog has name:", dog.name)
print("Cat has name:", cat.name)

print("Dog class:", type(dog))
print("Is dog an Animal?", isinstance(dog, Animal))
print("Is Dog a subclass of Animal?", issubclass(Dog, Animal))
````

# 5. Method Overriding

A child class can provide its own version of a method already defined in the parent class.

This is called **method overriding**.

````python
class Animal:
    def speak(self):
        return "Some animal sound"


class Dog(Animal):
    def speak(self):
        return "Woof!"


class Cat(Animal):
    def speak(self):
        return "Meow!"


print(Animal().speak())
print(Dog().speak())
print(Cat().speak())
````

This connects inheritance with **polymorphism**: different subclasses respond differently to the same method call.

````python
animals = [Dog(), Cat(), Animal()]

for animal in animals:
    print(animal.speak())
````

# 6. Extending the Parent Constructor with `super()`

Often a child class needs all parent attributes **plus additional attributes**.

Use `super()` to call the parent implementation.

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def summary(self):
        return f"{self.name} earns ${self.salary:,.2f}"


class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department

    def summary(self):
        base = super().summary()
        return f"{base} and manages the {self.department} department."


manager = Manager("Priya", 95000, "Data Science")

print(manager.name)
print(manager.salary)
print(manager.department)
print(manager.summary())
````

### Why use `super()`?

Instead of rewriting the parent's constructor:

```python
self.name = name
self.salary = salary
```

we reuse the existing implementation:

```python
super().__init__(name, salary)
```

Benefits:

- less duplicated code
- easier maintenance
- clearer relationship between parent and child classes
- safer refactoring

# 7. Types of Inheritance

Python supports several inheritance structures.

We will look at:

1. Single inheritance
2. Multilevel inheritance
3. Hierarchical inheritance
4. Multiple inheritance

## 7.1 Single Inheritance

One child inherits from one parent.

```text
Vehicle
   |
  Car
```

````python
class Vehicle:
    def move(self):
        return "Vehicle is moving"


class Car(Vehicle):
    def drive(self):
        return "Car is driving"


car = Car()

print(car.move())
print(car.drive())
````

## 7.2 Multilevel Inheritance

A class inherits from a class that already inherited from another class.

```text
LivingThing
    |
  Animal
    |
   Dog
```

````python
class LivingThing:
    def grow(self):
        return "Growing..."


class Animal(LivingThing):
    def eat(self):
        return "Eating..."


class Dog(Animal):
    def bark(self):
        return "Woof!"


dog = Dog()

print(dog.grow())
print(dog.eat())
print(dog.bark())
````

## 7.3 Hierarchical Inheritance

Several child classes inherit from the same parent.

```text
        Shape
       /     \
   Circle   Rectangle
```

````python
import math

class Shape:
    def describe(self):
        return "I am a shape."


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


circle = Circle(3)
rectangle = Rectangle(4, 5)

print(circle.describe())
print("Circle area:", round(circle.area(), 2))

print(rectangle.describe())
print("Rectangle area:", rectangle.area())
````

## 7.4 Multiple Inheritance

A class can inherit from more than one parent.

```text
Camera      Phone
    \       /
     SmartPhone
```

Multiple inheritance is powerful, but it should be used carefully because method lookup can become more complex.

````python
class Camera:
    def take_photo(self):
        return "Photo taken."


class Phone:
    def make_call(self):
        return "Calling..."


class SmartPhone(Camera, Phone):
    def browse_web(self):
        return "Opening browser..."


device = SmartPhone()

print(device.take_photo())
print(device.make_call())
print(device.browse_web())
````

# 8. Method Resolution Order (MRO)

When multiple parent classes contain methods with the same name, Python must decide which implementation to use.

Python follows the **Method Resolution Order (MRO)**.

````python
class A:
    def identify(self):
        return "Method from A"


class B(A):
    def identify(self):
        return "Method from B"


class C(A):
    def identify(self):
        return "Method from C"


class D(B, C):
    pass


obj = D()

print(obj.identify())
print(D.mro())
````

For `D(B, C)`, Python searches classes in an order similar to:

```text
D → B → C → A → object
```

You can inspect it with:

```python
D.mro()
```

or:

```python
D.__mro__
```

# 9. Worked Example: University People System

Now we will build a slightly more realistic hierarchy.

```text
Person
├── Student
└── Staff
    └── Lecturer
```

The shared data belongs in the highest sensible parent class.

````python
class Person:
    def __init__(self, name, email):
        self.name = name
        self.email = email

    def contact_card(self):
        return f"{self.name} <{self.email}>"


class Student(Person):
    def __init__(self, name, email, student_id):
        super().__init__(name, email)
        self.student_id = student_id

    def role(self):
        return "Student"


class Staff(Person):
    def __init__(self, name, email, staff_id):
        super().__init__(name, email)
        self.staff_id = staff_id

    def role(self):
        return "Staff"


class Lecturer(Staff):
    def __init__(self, name, email, staff_id, course):
        super().__init__(name, email, staff_id)
        self.course = course

    def role(self):
        return "Lecturer"

    def teach(self):
        return f"{self.name} is teaching {self.course}."


student = Student("Emily", "emily@example.edu", "S12345")
lecturer = Lecturer("Dr Chen", "chen@example.edu", "E900", "Intermediate Programming")

print(student.contact_card())
print(student.role())
print()

print(lecturer.contact_card())
print(lecturer.role())
print(lecturer.teach())
````

### Trace the `Lecturer` object

When this line runs:

```python
lecturer = Lecturer(...)
```

the constructor chain is:

```text
Lecturer.__init__()
        ↓
Staff.__init__()
        ↓
Person.__init__()
```

Each class contributes the information relevant to its level of the hierarchy.

````python
print("Lecturer MRO:")
for cls in Lecturer.mro():
    print(" -", cls.__name__)
````

# 10. Inherited Class Variables

````python
class Employee:
    organisation = "TechNova"

    def __init__(self, name):
        self.name = name


class Developer(Employee):
    pass


dev = Developer("Sam")

print(dev.name)
print(dev.organisation)
print(Developer.organisation)
````

# 11. Checking Inheritance Relationships

Two useful built-in functions are:

### `isinstance(object, Class)`
Checks whether an object belongs to a class **or one of its subclasses**.

### `issubclass(Child, Parent)`
Checks whether one class inherits from another.

````python
class Animal:
    pass

class Dog(Animal):
    pass

buddy = Dog()

print(isinstance(buddy, Dog))
print(isinstance(buddy, Animal))
print(issubclass(Dog, Animal))
print(issubclass(Animal, Dog))
````

# 12. Inheritance with Abstract Base Classes

Inheritance is often combined with abstraction.

A parent class can define a required interface while leaving the implementation to subclasses.

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2


class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height


shapes = [Square(4), Triangle(6, 3)]

for shape in shapes:
    print(type(shape).__name__, "area =", shape.area())
````

The abstract parent says:

> Every concrete `Shape` must provide an `area()` method.

Each child class decides **how** to calculate the area.

# 13. Inheritance vs Composition

Inheritance is best when there is a strong **is-a** relationship.

- `Dog` **is an** `Animal` → inheritance makes sense.
- `Manager` **is an** `Employee` → inheritance makes sense.

Composition is often better for **has-a** relationships.

- `Car` **has an** `Engine`.
- `Computer` **has a** `Processor`.
- `Course` **has a** `Teacher`.

Trying to use inheritance for a "has-a" relationship usually creates an unnatural design.

````python
class Engine:
    def start(self):
        return "Engine started."


class Car:
    def __init__(self):
        self.engine = Engine()  # Car HAS an Engine

    def start(self):
        return self.engine.start()


car = Car()
print(car.start())
````

### Rule of thumb

Before using inheritance, ask:

> **Can I naturally say "Child is a Parent"?**

If yes, inheritance may be appropriate.

If the relationship is instead:

> **"Object has another object"**

composition may be clearer.

# 14. Common Inheritance Mistakes

### Mistake 1 — Forgetting to initialise the parent

If the child defines its own `__init__`, the parent constructor is **not automatically called**.

Use:

```python
super().__init__(...)
```

when parent initialisation is required.

### Mistake 2 — Using inheritance only to avoid typing code

Inheritance should represent a meaningful relationship, not just code reuse.

### Mistake 3 — Building very deep class hierarchies

Deep inheritance trees are difficult to understand and maintain.

### Mistake 4 — Overusing multiple inheritance

Multiple inheritance is useful in some designs but can make method lookup less obvious.

### Mistake 5 — Overriding a method with incompatible behaviour

A subclass should generally preserve the expectations established by the parent interface.

````python
# Interactive widget version: one question at a time, four clickable options.

try:
    import ipywidgets as widgets
    from IPython.display import display, clear_output

    class InheritanceQuiz:
        def __init__(self, questions):
            self.questions = questions
            self.current = 0
            self.score = 0
            self.answered = False

            self.progress = widgets.HTML()
            self.question_text = widgets.HTML()

            self.options = widgets.RadioButtons(
                options=[],
                description="",
                disabled=False,
                layout=widgets.Layout(width="95%")
            )

            self.submit_button = widgets.Button(
                description="Submit Answer",
                button_style="primary",
                icon="check"
            )

            self.next_button = widgets.Button(
                description="Next Question",
                button_style="success",
                icon="arrow-right",
                disabled=True
            )

            self.feedback = widgets.HTML()
            self.score_display = widgets.HTML()

            self.submit_button.on_click(self.submit_answer)
            self.next_button.on_click(self.next_question)

            self.container = widgets.VBox([
                self.progress,
                self.question_text,
                self.options,
                widgets.HBox([self.submit_button, self.next_button]),
                self.feedback,
                self.score_display
            ])

            self.show_question()

        def show_question(self):
            q = self.questions[self.current]
            self.answered = False

            self.progress.value = (
                f"<b>Question {self.current + 1} of {len(self.questions)}</b>"
            )
            self.question_text.value = (
                f"<h3>{q['question']}</h3>"
            )

            # Four clickable options
            self.options.options = [
                (f"{chr(65+i)}. {option}", i)
                for i, option in enumerate(q["options"])
            ]
            self.options.value = None
            self.options.disabled = False

            self.submit_button.disabled = False
            self.next_button.disabled = True
            self.feedback.value = ""
            self.update_score()

        def submit_answer(self, _):
            if self.answered:
                return

            if self.options.value is None:
                self.feedback.value = (
                    "<span style='color:#b26a00'><b>Please select one option first.</b></span>"
                )
                return

            q = self.questions[self.current]
            selected = self.options.value
            correct = q["answer"]

            self.answered = True
            self.options.disabled = True
            self.submit_button.disabled = True

            if selected == correct:
                self.score += 1
                self.feedback.value = (
                    "<span style='color:green'><b>Correct!</b></span>"
                    f"<br>{q['explanation']}"
                )
            else:
                correct_letter = chr(65 + correct)
                correct_text = q["options"][correct]
                self.feedback.value = (
                    "<span style='color:crimson'><b>Incorrect.</b></span>"
                    f"<br>Correct answer: <b>{correct_letter}. {correct_text}</b>"
                    f"<br>{q['explanation']}"
                )

            if self.current == len(self.questions) - 1:
                self.next_button.description = "Show Final Result"
            else:
                self.next_button.description = "Next Question"

            self.next_button.disabled = False
            self.update_score()

        def next_question(self, _):
            if not self.answered:
                return

            if self.current < len(self.questions) - 1:
                self.current += 1
                self.show_question()
            else:
                self.show_final_result()

        def update_score(self):
            self.score_display.value = (
                f"<b>Current score:</b> {self.score} / {len(self.questions)}"
            )

        def show_final_result(self):
            percentage = (self.score / len(self.questions)) * 100

            if percentage >= 80:
                message = "Excellent work — you have a strong understanding of inheritance."
            elif percentage >= 60:
                message = "Good work — review a few concepts and try again."
            else:
                message = "Review the inheritance examples above, then retake the quiz."

            self.progress.value = "<b>Quiz complete</b>"
            self.question_text.value = (
                f"<h2>Final Score: {self.score}/{len(self.questions)} "
                f"({percentage:.0f}%)</h2>"
            )
            self.options.options = []
            self.feedback.value = f"<p>{message}</p>"
            self.submit_button.layout.display = "none"
            self.next_button.layout.display = "none"
            self.update_score()


    inheritance_quiz = InheritanceQuiz(quiz_questions)
    display(inheritance_quiz.container)

except ImportError:
    print(
        "ipywidgets is not installed in this environment. "
        "Run `%pip install ipywidgets` and then rerun this cell, "
        "or use the text-based quiz below."
    )
````

## Text-Based Quiz Fallback

Use this version if widget controls are unavailable.

It still presents **one question at a time** with **four options**.
Enter `A`, `B`, `C`, or `D`.

````python
def run_text_quiz(questions):
    score = 0

    for index, q in enumerate(questions, start=1):
        print("\n" + "=" * 70)
        print(f"Question {index} of {len(questions)}")
        print(q["question"])
        print()

        for i, option in enumerate(q["options"]):
            print(f"{chr(65+i)}. {option}")

        while True:
            answer = input("\nYour answer (A/B/C/D): ").strip().upper()
            if answer in {"A", "B", "C", "D"}:
                break
            print("Please enter A, B, C, or D.")

        selected = ord(answer) - ord("A")

        if selected == q["answer"]:
            score += 1
            print("Correct!")
        else:
            correct = q["answer"]
            print(
                f"Incorrect. Correct answer: "
                f"{chr(65+correct)}. {q['options'][correct]}"
            )

        print(q["explanation"])
        print(f"Score: {score}/{index}")

    percentage = (score / len(questions)) * 100

    print("\n" + "=" * 70)
    print("QUIZ COMPLETE")
    print(f"Final score: {score}/{len(questions)} ({percentage:.0f}%)")


# Run when you are ready:
# run_text_quiz(quiz_questions)
````
