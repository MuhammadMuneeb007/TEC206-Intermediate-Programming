## Big idea

Python treats functions as **objects**.

That means a function can be:

- assigned to a variable,
- passed as an argument,
- returned from another function,
- stored inside data structures.

This ability is the foundation of **higher-order functions**.

# 1. Lambda Functions

A **lambda function** is a small function written without using the `def` keyword.

It is also commonly called an **anonymous function** because it can be created without giving it a traditional function name.

Lambda functions are useful when we need a short function, especially when that function will only be used once or passed directly into another function.

### Generic syntax

```python
lambda arguments: expression
```

For example:

```python
lambda x: x * 2
```

This function:

- receives `x`,
- calculates `x * 2`,
- automatically returns the result.

A lambda normally contains **one expression**, not a sequence of statements.

## 1.1 Regular function versus lambda function

First, consider a regular function.

````python
def square(x):
    return x ** 2

square(5)
````

The same operation can be written using a lambda function:

````python
square_lambda = lambda x: x ** 2

square_lambda(5)
````

Both functions produce the same result.

The important difference is mostly **syntax and intended use**.

| Regular function | Lambda function |
|---|---|
| Uses `def` | Uses `lambda` |
| Usually has a name | Can be anonymous |
| Can contain multiple statements | Normally contains one expression |
| Good for larger/reusable logic | Good for short/simple operations |
| Uses `return` explicitly | Returns the expression automatically |

## 1.2 Understanding the lambda syntax

Consider:

```python
lambda x: x + 10
```

We can divide it into three parts:

```text
lambda    x    :    x + 10
  |       |           |
keyword argument   expression
```

The expression is evaluated and returned automatically.

````python
add_ten = lambda x: x + 10

print(add_ten(5))
print(add_ten(20))
````

## 1.3 Lambda functions with multiple arguments

A lambda function can receive more than one argument.

````python
add = lambda x, y: x + y

print(add(3, 4))
print(add(10, 25))
````

````python
multiply = lambda a, b: a * b

multiply(6, 7)
````

## 1.4 Lambda functions with conditions

A conditional expression can also be used inside a lambda function.

The syntax is:

```python
value_if_true if condition else value_if_false
```

````python
check_even = lambda x: "Even" if x % 2 == 0 else "Odd"

print(check_even(8))
print(check_even(7))
````

## 1.5 When should we use lambda?

Lambda functions are useful when:

- the logic is short,
- the function is used only once,
- a function must be passed into another function,
- we are using functions such as `map()`, `filter()`, or `sorted()`.

For example:

````python
students = [
    ("Aisha", 82),
    ("Ben", 71),
    ("Carlos", 95),
    ("Dina", 88)
]

sorted_students = sorted(students, key=lambda student: student[1])

sorted_students
````

Here, `sorted()` needs to know **what part of each item should be used for sorting**.

The lambda:

```python
lambda student: student[1]
```

tells Python:

> "For each student tuple, use the second value — the mark — as the sorting key."

This is a common practical use of lambda functions.

# 2. Functions as Objects

Before learning higher-order functions, we need one important idea:

> In Python, a function can be treated like a value.

Consider this function:

````python
def greet(name):
    return f"Hello, {name}!"

greet("Alex")
````

We can assign the function itself to another variable.

Notice that we write `greet`, not `greet()`.

````python
another_name = greet

print(another_name("Maya"))
````

Why?

- `greet` refers to the function itself.
- `greet()` calls the function.

This distinction becomes very important when functions are passed into other functions.

# 3. Higher-Order Functions

A **higher-order function** is a function that does at least one of the following:

1. **Accepts another function as an argument**
2. **Returns another function**

Sometimes it does both.

Higher-order functions are useful because they allow us to separate:

- **what operation should happen**
from
- **how or when that operation should be applied**

This makes code more flexible and reusable.

## 3.1 Passing a function as an argument

Let us create two simple functions.

````python
def square(x):
    return x ** 2

def double(x):
    return x * 2
````

Now we create a function that receives another function.

````python
def apply_operation(function, value):
    return function(value)

print(apply_operation(square, 5))
print(apply_operation(double, 5))
````

`apply_operation()` is a **higher-order function** because one of its arguments is itself a function.

For:

```python
apply_operation(square, 5)
```

Python effectively performs:

```python
square(5)
```

For:

```python
apply_operation(double, 5)
```

Python effectively performs:

```python
double(5)
```

The important benefit is that `apply_operation()` does not need to know the exact operation in advance.

We can change its behaviour simply by passing in a different function.

## 3.2 Passing a lambda function as an argument

Instead of defining a separate function, we can pass a lambda directly.

````python
result = apply_operation(lambda x: x ** 3, 4)

result
````

Here:

```python
lambda x: x ** 3
```

is the function being passed into the higher-order function.

### Important distinction

A lambda function is **not automatically a higher-order function**.

A lambda is simply a way of creating a function.

In this expression:

```python
apply_operation(lambda x: x ** 3, 4)
```

- `lambda x: x ** 3` is the function being passed,
- `apply_operation()` is the higher-order function.

# 4. Returning a Function from Another Function

A function can also create and return another function.

Consider this example:

````python
def make_multiplier(number):
    def multiply(value):
        return value * number

    return multiply
````

Now we can create specialised functions.

````python
double = make_multiplier(2)
triple = make_multiplier(3)
multiply_by_ten = make_multiplier(10)

print(double(5))
print(triple(5))
print(multiply_by_ten(5))
````

`make_multiplier()` is a higher-order function because it **returns another function**.

This is useful when we want to create customised behaviour.

For example:

```python
double = make_multiplier(2)
```

creates a function that remembers the value `2`.

Later:

```python
double(10)
```

returns `20`.

This is also an introduction to an important concept called a **closure**: the returned function remembers information from the environment in which it was created.

# 5. Why Are Higher-Order Functions Useful?

Suppose we have some numbers:

````python
numbers = [1, 2, 3, 4, 5]
````

Without higher-order functions, we might repeatedly write loops.

For example, to square every value:

````python
squared = []

for number in numbers:
    squared.append(number ** 2)

squared
````

To double every value, we would write another very similar loop:

````python
doubled = []

for number in numbers:
    doubled.append(number * 2)

doubled
````

The structure of the loop is almost identical.

Only the **operation** changes.

Higher-order functions allow us to separate the repeated processing pattern from the operation being performed.

This can make code:

- more reusable,
- more expressive,
- easier to customise,
- more compact,
- easier to combine into data-processing pipelines.

# 6. `map()`

`map()` is a built-in higher-order function.

It applies a function to every item in an iterable.

### Generic syntax

```python
map(function, iterable)
```

Conceptually:

```text
input values
     |
     v
function applied to every value
     |
     v
transformed values
```

````python
numbers = [1, 2, 3, 4, 5]

def square(x):
    return x ** 2

result = map(square, numbers)

list(result)
````

The same code can be written using a lambda:

````python
numbers = [1, 2, 3, 4, 5]

result = map(lambda x: x ** 2, numbers)

list(result)
````

### What happened?

`map()` effectively performed:

```python
square(1)
square(2)
square(3)
square(4)
square(5)
```

and generated:

```python
[1, 4, 9, 16, 25]
```

### Typical uses of `map()`

`map()` is useful when we want to **transform every item**.

Examples include:

- converting strings to numbers,
- converting temperatures,
- normalising values,
- cleaning text,
- applying mathematical transformations,
- extracting or transforming fields in data.

## 6.1 Example: Convert Celsius to Fahrenheit

````python
celsius = [0, 10, 20, 30, 40]

fahrenheit = list(
    map(lambda c: (c * 9 / 5) + 32, celsius)
)

fahrenheit
````

## 6.2 Example: Clean text

````python
names = ["  alice ", "BOB", "  Charlie  "]

cleaned_names = list(
    map(lambda name: name.strip().title(), names)
)

cleaned_names
````

# 7. `filter()`

`filter()` is another higher-order function.

Instead of transforming every item, `filter()` **keeps only the items that satisfy a condition**.

### Generic syntax

```python
filter(function, iterable)
```

The function should return something that evaluates to either:

- `True` → keep the item
- `False` → remove the item

````python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even_numbers = list(
    filter(lambda x: x % 2 == 0, numbers)
)

even_numbers
````

Conceptually:

```text
1 -> False -> remove
2 -> True  -> keep
3 -> False -> remove
4 -> True  -> keep
5 -> False -> remove
6 -> True  -> keep
```

## 7.1 Example: Filter passing marks

````python
marks = [45, 67, 82, 39, 91, 55, 30]

passing_marks = list(
    filter(lambda mark: mark >= 50, marks)
)

passing_marks
````

## 7.2 Example: Filter names by length

````python
names = ["Ana", "Mohammed", "Li", "Charlotte", "Ben"]

long_names = list(
    filter(lambda name: len(name) >= 5, names)
)

long_names
````

### Typical uses of `filter()`

`filter()` is useful when we want to select only records that satisfy some rule.

Examples include:

- students who passed,
- products below a certain price,
- valid measurements,
- active users,
- positive values,
- genes meeting a threshold,
- records matching a condition.

# 8. `reduce()`

`reduce()` is different from `map()` and `filter()`.

Instead of producing many output values, `reduce()` repeatedly combines values until a **single result** remains.

`reduce()` is available from Python's `functools` module.

### Generic syntax

```python
reduce(function, iterable)
```

or:

```python
reduce(function, iterable, initial_value)
```

````python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

total = reduce(lambda x, y: x + y, numbers)

total
````

Conceptually, the operation is similar to:

```text
1 + 2 = 3
3 + 3 = 6
6 + 4 = 10
10 + 5 = 15
```

Eventually only one value remains:

```text
15
```

## 8.1 Example: Product of all numbers

````python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

product = reduce(lambda x, y: x * y, numbers)

product
````

## 8.2 Example: Find the maximum

````python
numbers = [8, 2, 17, 4, 11]

maximum = reduce(
    lambda x, y: x if x > y else y,
    numbers
)

maximum
````

### Typical uses of `reduce()`

`reduce()` can be useful for:

- summing values,
- multiplying values,
- combining records,
- accumulating results,
- finding a maximum or minimum,
- building a single result from many items.

However, in real Python code, clearer built-in alternatives are often preferred when available.

For example:

```python
sum(numbers)
max(numbers)
min(numbers)
```

are usually clearer than using `reduce()` for those specific tasks.

# 9. `map()` vs `filter()` vs `reduce()`

The easiest way to remember them is:

| Function | Main purpose | Input | Output idea |
|---|---|---|---|
| `map()` | Transform | many values | many transformed values |
| `filter()` | Select | many values | some of the original values |
| `reduce()` | Combine | many values | usually one final value |

### Mental model

```text
MAP
[1, 2, 3] -> transform each -> [2, 4, 6]

FILTER
[1, 2, 3, 4] -> keep selected -> [2, 4]

REDUCE
[1, 2, 3, 4] -> combine -> 10
```

# 10. Combining `map()`, `filter()`, and `reduce()`

These functions can be combined to build a data-processing pipeline.

Suppose we want to:

1. start with several numbers,
2. keep only the even numbers,
3. square the remaining values,
4. add all the squared values together.

````python
from functools import reduce

numbers = [1, 2, 3, 4, 5, 6]

even_numbers = filter(lambda x: x % 2 == 0, numbers)

squared_even_numbers = map(lambda x: x ** 2, even_numbers)

total = reduce(lambda x, y: x + y, squared_even_numbers)

total
````

The pipeline is:

```text
[1, 2, 3, 4, 5, 6]

        filter
           |
           v

      [2, 4, 6]

          map
           |
           v

      [4, 16, 36]

         reduce
           |
           v

           56
```

# 11. A More Realistic Example: Student Results

Suppose we have student records.

````python
students = [
    {"name": "Aisha", "mark": 82},
    {"name": "Ben", "mark": 43},
    {"name": "Carlos", "mark": 91},
    {"name": "Dina", "mark": 55},
    {"name": "Ethan", "mark": 38}
]

students
````

First, keep only students who passed.

````python
passed = list(
    filter(lambda student: student["mark"] >= 50, students)
)

passed
````

Now extract their names.

````python
passed_names = list(
    map(lambda student: student["name"], passed)
)

passed_names
````

Now calculate the total mark for the students who passed.

````python
from functools import reduce

total_mark = reduce(
    lambda total, student: total + student["mark"],
    passed,
    0
)

total_mark
````

This illustrates how higher-order functions can be used for common data-processing tasks:

- `filter()` selects the records we care about,
- `map()` transforms or extracts information,
- `reduce()` combines information into one result.

# 12. Higher-Order Functions That Return Functions

Higher-order functions are not limited to `map()`, `filter()`, and `reduce()`.

They can also be used to create specialised functions.

````python
def power_maker(power):
    return lambda x: x ** power

square = power_maker(2)
cube = power_maker(3)

print(square(5))
print(cube(5))
````

This is useful because `power_maker()` allows us to create a family of related functions.

```python
square = power_maker(2)
cube = power_maker(3)
```

The returned functions remember the value of `power`.

This pattern is useful in:

- configuration,
- callbacks,
- decorators,
- event handling,
- reusable validation rules,
- custom transformations,
- machine-learning and data-processing pipelines.

# 13. Lambda Functions Are Not Always Better

Consider this:

````python
calculate = lambda x: ((x ** 2 + 3 * x - 5) / (x + 1)) if x != -1 else None
````

Although this is valid, a regular function may be easier to understand:

````python
def calculate(x):
    if x == -1:
        return None

    numerator = x ** 2 + 3 * x - 5
    denominator = x + 1

    return numerator / denominator
````

### Rule of thumb

Use `lambda` for short, simple operations.

Use `def` when the logic:

- contains several steps,
- needs comments or documentation,
- will be reused,
- becomes difficult to read in one line.

Readable code is usually more important than shorter code.

# 14. Pythonic Alternatives: List Comprehensions

In Python, `map()` and `filter()` are sometimes replaced by list comprehensions because many programmers find them easier to read.

### `map()` style

````python
numbers = [1, 2, 3, 4, 5]

list(map(lambda x: x ** 2, numbers))
````

### List comprehension style

````python
[x ** 2 for x in numbers]
````

### `filter()` style

````python
list(filter(lambda x: x % 2 == 0, numbers))
````

### List comprehension style

````python
[x for x in numbers if x % 2 == 0]
````

Neither style is universally "correct".

The important learning goal is to understand that `map()` and `filter()` are examples of **higher-order functions** because they receive functions as arguments.
