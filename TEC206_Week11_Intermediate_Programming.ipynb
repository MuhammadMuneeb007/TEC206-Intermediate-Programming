# 2. What Is a Set?

A **set** is a collection that:

1. stores **unique** elements;
2. is **unordered**;
3. is **mutable** — we can add and remove elements;
4. supports mathematical set operations.

Consider this list:

```python
music_genres = ["rock", "pop", "rock", "jazz", "pop", "hip-hop"]
```

The list keeps every occurrence.

A set keeps only the distinct values:

```python
{"rock", "pop", "jazz", "hip-hop"}
```

This makes sets very useful for **deduplication** and **group comparison**.

````python
music_genres = ["rock", "pop", "rock", "jazz", "pop", "hip-hop"]
unique_genres = set(music_genres)

print("Original list:", music_genres)
print("Unique genres:", unique_genres)
print("Number of entries in list:", len(music_genres))
print("Number of unique genres:", len(unique_genres))
````

### Key observation

`set(...)` removed repeated values automatically.

However, the order may look different each time or on different systems.

If you need a predictable display order, you can sort the set **for display**:

```python
sorted(unique_genres)
```

`sorted(...)` returns a **list**.

````python
print("Sorted for display:", sorted(unique_genres))
print("Type of unique_genres:", type(unique_genres))
print("Type of sorted(unique_genres):", type(sorted(unique_genres)))
````

# 6. Removing Elements

Python provides several ways to remove values from a set.

The most important distinction is between `remove()` and `discard()`.

## 6.1 `remove()` — Remove a Specific Element

`remove(value)` deletes the value.

If the value is not present, Python raises a `KeyError`.

````python
courses = {"Python", "Java", "C++"}
courses.remove("Java")
print(courses)
````

````python
courses = {"Python", "Java", "C++"}

try:
    courses.remove("R")
except KeyError:
    print("KeyError: 'R' was not present in the set.")
````

## 6.2 `discard()` — Remove Safely

`discard(value)` removes the value if it exists.

If it does not exist, nothing happens.

````python
courses = {"Python", "Java", "C++"}

courses.discard("Java")
courses.discard("R")     # no error

print(courses)
````

## 6.3 `pop()` — Remove an Arbitrary Element

Because sets are unordered, `.pop()` removes an **arbitrary** element.

Do not assume it removes the "first" or "last" item.

````python
letters = {"A", "B", "C", "D"}
removed = letters.pop()

print("Removed:", removed)
print("Remaining:", letters)
````

## 6.4 `clear()` — Remove Everything

````python
temporary_data = {10, 20, 30}
temporary_data.clear()

print(temporary_data)
print("Is it empty?", len(temporary_data) == 0)
````

### Removal summary

| Method | If element exists | If element is missing |
|---|---|---|
| `remove(x)` | Removes it | Raises `KeyError` |
| `discard(x)` | Removes it | Does nothing |
| `pop()` | Removes an arbitrary element | Raises `KeyError` if set is empty |
| `clear()` | Removes all elements | Leaves an empty set |

# 10. A Reusable Venn Diagram Function

The function below draws two overlapping circles and places:

- values found only in `A`;
- values found in both `A` and `B`;
- values found only in `B`.

It uses only `matplotlib`.

````python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

def draw_venn(A, B, label_a="A", label_b="B", title="Two-Set Venn Diagram"):
    only_a = A - B
    both = A & B
    only_b = B - A

    fig, ax = plt.subplots(figsize=(8, 5))

    left_circle = Circle((0.42, 0.5), 0.28, fill=False, linewidth=2)
    right_circle = Circle((0.58, 0.5), 0.28, fill=False, linewidth=2)

    ax.add_patch(left_circle)
    ax.add_patch(right_circle)

    ax.text(0.26, 0.82, label_a, fontsize=14, ha="center")
    ax.text(0.74, 0.82, label_b, fontsize=14, ha="center")

    ax.text(0.30, 0.50, "\n".join(map(str, sorted(only_a))) or "∅",
            fontsize=12, ha="center", va="center")
    ax.text(0.50, 0.50, "\n".join(map(str, sorted(both))) or "∅",
            fontsize=12, ha="center", va="center")
    ax.text(0.70, 0.50, "\n".join(map(str, sorted(only_b))) or "∅",
            fontsize=12, ha="center", va="center")

    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_aspect("equal")
    ax.axis("off")
    ax.set_title(title)
    plt.show()
````

````python
draw_venn(
    A,
    B,
    label_a="A: Even",
    label_b="B: Prime",
    title="Even Numbers and Prime Numbers"
)
````

# 11. Union — Everything in A or B

The **union** contains every element that appears in `A`, `B`, or both.

Python gives us two equivalent forms:

```python
A | B
A.union(B)
```

````python
union_result = A | B

print("A:", A)
print("B:", B)
print("A ∪ B:", union_result)
print("Using .union():", A.union(B))
````

### Real-life interpretation

Imagine:

- `A` = students enrolled in Python;
- `B` = students enrolled in Java.

`A | B` gives **all students enrolled in at least one of the two courses**.

# 12. Intersection — Common Elements

The **intersection** contains elements found in **both** sets.

Equivalent forms:

```python
A & B
A.intersection(B)
```

````python
intersection_result = A & B

print("A:", A)
print("B:", B)
print("A ∩ B:", intersection_result)
print("Using .intersection():", A.intersection(B))
````

For our even/prime example, the common value is `2`.

### Real-life interpretation

If:

- `A` = users who bought Product A;
- `B` = users who bought Product B;

then `A & B` gives the customers who bought **both products**.

# 13. Difference — In One Set but Not the Other

This operation is **directional**.

`A - B` means:

> elements that are in `A` but **not** in `B`.

`B - A` means something different:

> elements that are in `B` but **not** in `A`.

````python
print("A - B:", A - B)
print("B - A:", B - A)
print("Using .difference():", A.difference(B))
````

### Very important

Difference is **not symmetric**:

```python
A - B != B - A
```

in general.

### Real-life example

Suppose:

- `A` = customers who subscribed last month;
- `B` = customers who subscribed this month.

Then:

- `A - B` = customers who were present last month but not this month;
- `B - A` = new customers this month.

# 14. Symmetric Difference — In Either Set, but Not Both

The **symmetric difference** contains values that are in exactly one set.

It excludes the overlap.

Equivalent forms:

```python
A ^ B
A.symmetric_difference(B)
```

````python
symmetric_result = A ^ B

print("A:", A)
print("B:", B)
print("A △ B:", symmetric_result)
print("Using .symmetric_difference():", A.symmetric_difference(B))
````

A useful way to understand it is:

```python
(A - B) | (B - A)
```

That means:

- everything unique to `A`;
- plus everything unique to `B`;
- but nothing shared by both.

````python
manual_symmetric = (A - B) | (B - A)

print("Using ^ :", A ^ B)
print("Manual :", manual_symmetric)
print("Same result?", (A ^ B) == manual_symmetric)
````

# 15. The Four Main Set Operations — One Table

| Operation | Symbol | Python | Meaning |
|---|---:|---|---|
| Union | `A ∪ B` | `A | B` | Everything in A or B |
| Intersection | `A ∩ B` | `A & B` | Elements common to both |
| Difference | `A − B` | `A - B` | Elements in A but not B |
| Symmetric difference | `A △ B` | `A ^ B` | Elements in either set but not both |

````python
print("A =", A)
print("B =", B)
print()
print("Union                A | B =", A | B)
print("Intersection         A & B =", A & B)
print("Difference           A - B =", A - B)
print("Difference           B - A =", B - A)
print("Symmetric difference A ^ B =", A ^ B)
````

# 16. Venn Diagram Thinking

For two sets `A` and `B`, imagine three visible regions:

```text
        A                    B
   ┌──────────┐        ┌──────────┐
   │ A only   │────────│ B only   │
   │          │  both  │          │
   └──────────┘        └──────────┘
```

- **Union** = A only + both + B only
- **Intersection** = both
- **A − B** = A only
- **B − A** = B only
- **Symmetric difference** = A only + B only

````python
draw_venn(
    {"Python", "SQL", "Git", "Linux"},
    {"Python", "Java", "Git", "Docker"},
    label_a="Data Team",
    label_b="Software Team",
    title="Venn Diagram: Team Skills"
)
````

## 16.1 Operation-by-Operation Venn Diagrams

The next visualisations use the same three Venn regions:

- **A only**
- **A and B**
- **B only**

A `✓` marks the region selected by each operation.

````python
def draw_operation_venn(A, B, operation, label_a="A", label_b="B"):
    only_a = A - B
    both = A & B
    only_b = B - A

    selected = {
        "union": (True, True, True),
        "intersection": (False, True, False),
        "a-b": (True, False, False),
        "b-a": (False, False, True),
        "symmetric": (True, False, True),
    }

    if operation not in selected:
        raise ValueError("Unknown operation.")

    select_a, select_both, select_b = selected[operation]

    titles = {
        "union": "Union: A | B — everything in either set",
        "intersection": "Intersection: A & B — common to both",
        "a-b": "Difference: A - B — in A but not B",
        "b-a": "Difference: B - A — in B but not A",
        "symmetric": "Symmetric Difference: A ^ B — in exactly one set",
    }

    fig, ax = plt.subplots(figsize=(8, 5))

    ax.add_patch(Circle((0.42, 0.5), 0.28, fill=False, linewidth=2))
    ax.add_patch(Circle((0.58, 0.5), 0.28, fill=False, linewidth=2))

    ax.text(0.26, 0.82, label_a, fontsize=14, ha="center")
    ax.text(0.74, 0.82, label_b, fontsize=14, ha="center")

    def region_text(values, chosen):
        values_text = ", ".join(map(str, sorted(values))) if values else "∅"
        return ("✓\n" if chosen else "") + values_text

    ax.text(0.30, 0.50, region_text(only_a, select_a),
            fontsize=12, ha="center", va="center")
    ax.text(0.50, 0.50, region_text(both, select_both),
            fontsize=12, ha="center", va="center")
    ax.text(0.70, 0.50, region_text(only_b, select_b),
            fontsize=12, ha="center", va="center")

    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_aspect("equal")
    ax.axis("off")
    ax.set_title(titles[operation])
    plt.show()
````

### Union Venn diagram — `A | B`

Every Venn region is included.

````python
draw_operation_venn(A, B, "union", "A: Even", "B: Prime")
print("Result:", A | B)
````

### Intersection Venn diagram — `A & B`

Only the overlap is included.

````python
draw_operation_venn(A, B, "intersection", "A: Even", "B: Prime")
print("Result:", A & B)
````

### Difference Venn diagram — `A - B`

Only the `A`-only region is included.

This is why **direction matters**.

````python
draw_operation_venn(A, B, "a-b", "A: Even", "B: Prime")
print("A - B:", A - B)
print("B - A:", B - A)
````

### Symmetric difference Venn diagram — `A ^ B`

The two outside regions are included, but the overlap is excluded.

````python
draw_operation_venn(A, B, "symmetric", "A: Even", "B: Prime")
print("Result:", A ^ B)
````

# 21. Subsets and Supersets

A set `A` is a **subset** of `B` if every element of `A` is also in `B`.

Python provides:

```python
A <= B
A.issubset(B)
```

A set `B` is a **superset** of `A` if it contains every element of `A`:

```python
B >= A
B.issuperset(A)
```

````python
required = {"read", "comment"}
user = {"read", "comment", "download"}

print("required <= user:", required <= user)
print("required.issubset(user):", required.issubset(user))

print("user >= required:", user >= required)
print("user.issuperset(required):", user.issuperset(required))
````

## 21.1 Proper Subset and Proper Superset

`A < B` means:

- every element of `A` is in `B`;
- and `A` is not equal to `B`.

Similarly, `A > B` checks for a proper superset.

````python
small = {1, 2}
large = {1, 2, 3}

print("small <= large:", small <= large)
print("small < large:", small < large)
print("large > small:", large > small)
````

# 23. Set Comprehensions

A set comprehension has a similar structure to a list comprehension:

```python
{expression for item in iterable if condition}
```

The difference is that duplicate results are automatically removed.

````python
squares = {x * x for x in range(-5, 6)}
print(squares)
````

Notice that both `3**2` and `(-3)**2` produce `9`, but the set stores `9` only once.

````python
even_squares = {x * x for x in range(1, 11) if x % 2 == 0}
print(even_squares)
````

# 24. What Can Be Stored in a Set?

Set elements must be **hashable**.

Common hashable values include:

- integers;
- floats;
- strings;
- tuples containing hashable values;
- `frozenset`.

Mutable values such as lists, dictionaries, and normal sets cannot be set elements.

````python
valid = {
    10,
    3.14,
    "Python",
    (1, 2),
    frozenset({"A", "B"})
}

print(valid)
````

````python
try:
    invalid = {[1, 2], [3, 4]}
except TypeError as error:
    print("TypeError:", error)
````

### Why?

A set depends on stable hash values.

If a mutable list could change after insertion, its identity for hashing could become inconsistent.

Therefore, mutable lists are not valid set elements.

# 25. `frozenset` — An Immutable Set

A `frozenset` behaves like a set for most read-only set operations, but it cannot be modified.

````python
frozen = frozenset({"Python", "Java", "C++"})

print(frozen)
print("Python" in frozen)
````

````python
try:
    frozen.add("R")
except AttributeError as error:
    print("AttributeError:", error)
````

A `frozenset` can itself be stored inside another set because it is immutable and hashable.

````python
groups = {
    frozenset({"A", "B"}),
    frozenset({"C", "D"})
}

print(groups)
````

# 26. Sets vs Lists — Choosing the Right Collection

| Requirement | Usually choose |
|---|---|
| Preserve insertion/order and duplicates | `list` |
| Access by numeric index | `list` |
| Store unique values | `set` |
| Fast membership checks | `set` |
| Union/intersection/difference | `set` |
| Need key → value mapping | `dict` |
| Immutable unique collection | `frozenset` |

A set is **not** automatically better than a list.

Choose based on what your program needs.

# 27. Simple Membership Performance Demonstration

The following example compares repeated membership checks.

Exact timing differs by machine, but the set is normally much faster for membership lookup in a large collection.

````python
import time

N = 200_000
numbers_list = list(range(N))
numbers_set = set(numbers_list)
target = N - 1

start = time.perf_counter()
for _ in range(200):
    target in numbers_list
list_time = time.perf_counter() - start

start = time.perf_counter()
for _ in range(200):
    target in numbers_set
set_time = time.perf_counter() - start

print(f"List membership time: {list_time:.6f} seconds")
print(f"Set membership time : {set_time:.6f} seconds")
print("Your exact times will vary.")
````

Find:

7. whether `{"Ben", "Dina"}` is a subset of Workshop A;

````python
all_attendees = workshop_a | workshop_b
both = workshop_a & workshop_b
only_a = workshop_a - workshop_b
only_b = workshop_b - workshop_a
exactly_one = workshop_a ^ workshop_b

print("1. All:", all_attendees)
print("2. Both:", both)
print("3. Only A:", only_a)
print("4. Only B:", only_b)
print("5. Exactly one:", exactly_one)
print("6. Unique count:", len(all_attendees))
print("7. {'Ben', 'Dina'} subset of A?", {"Ben", "Dina"} <= workshop_a)
print("8. Disjoint?", workshop_a.isdisjoint(workshop_b))
````

# 31. Common Mistakes

## Mistake 1 — Using `{}` for an empty set

```python
x = {}      # dictionary
x = set()   # set
```

## Mistake 2 — Expecting sets to preserve a meaningful order

Do not write code that relies on the displayed position of set elements.

## Mistake 3 — Confusing difference with symmetric difference

```python
A - B   # only A
A ^ B   # A only + B only
```

## Mistake 4 — Using `remove()` when the value might be missing

Use `discard()` when absence is acceptable.

## Mistake 5 — Trying to store a list inside a set

Lists are mutable and unhashable.

## Mistake 6 — Expecting duplicates to remain

A set stores each unique value once.

## Mistake 7 — Indexing a set

This is invalid:

```python
my_set[0]
```

Sets do not provide numeric indexing.

# 35. Final Summary

A Python set is a powerful collection for **unique data** and **group relationships**.

The most important ideas from this lecture are:

- sets contain unique elements;
- sets are unordered;
- use `set()` for an empty set;
- `add()` adds one item;
- `update()` adds multiple items;
- `remove()` raises an error when the item is missing;
- `discard()` safely ignores a missing item;
- `|` gives the union;
- `&` gives the intersection;
- `-` gives directional difference;
- `^` gives symmetric difference;
- subset/superset operations test containment;
- `isdisjoint()` checks whether there is no overlap;
- sets are highly useful for duplicate removal and fast membership checks;
- Venn diagrams provide a strong mental model for set operations.

> **Remember:** when a problem asks about **unique values, overlap, membership, or group comparison**, think about using a set.
