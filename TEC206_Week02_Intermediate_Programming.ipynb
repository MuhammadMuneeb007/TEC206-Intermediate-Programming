# 1. Positional arguments

Positional arguments are matched to parameters according to their order.

````python
def introduce_student(name, course):
    return f"{name} is studying {course}."


print(introduce_student("Aisha", "Python"))
print(introduce_student("Noah", "Data Science"))
````

Changing the order changes the meaning because Python matches each value by position.

````python
print(introduce_student("Python", "Aisha"))  # Valid Python, but the meaning is wrong
````

# 2. Keyword arguments

Keyword arguments are matched using parameter names. Their order in the function call does not matter.

````python
def describe_book(title, author, year):
    return f"'{title}' was written by {author} and published in {year}."


print(describe_book(
    author="George Orwell",
    year=1949,
    title="Nineteen Eighty-Four",
))
````

Keyword arguments often make function calls easier to read.

# 3. Default arguments

A default argument is used when the caller does not provide a value.

````python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"


print(greet("Mina"))
print(greet("Mina", "Welcome"))
print(greet(name="Mina", greeting="Good morning"))
````

A parameter without a default value should normally appear before parameters with default values.

````python
def register_student(name, course="Python", active=True):
    return {
        "name": name,
        "course": course,
        "active": active,
    }


print(register_student("Sam"))
print(register_student("Sam", course="Databases"))
print(register_student("Sam", active=False))
````

# 4. Variable-length positional arguments: `*args`

`*args` collects extra positional arguments into a tuple.

The name `args` is conventional, but the `*` is the important part.

````python
def show_values(*args):
    print("Type:", type(args))
    print("Values:", args)


show_values(10, 20, 30)
show_values("Python", "SQL", "Git")
````

## 4.1 Calculate a sum from any number of values

````python
def calculate_sum(*numbers):
    return sum(numbers)


print(calculate_sum(1, 2, 3))
print(calculate_sum(10, 20, 30, 40))
print(calculate_sum())  # sum of an empty collection is 0
````

## 4.2 Pass multiple strings

````python
def join_words(*words, separator=" "):
    return separator.join(words)


print(join_words("Python", "is", "flexible"))
print(join_words("red", "green", "blue", separator=" | "))
````

Here, `words` collects all positional strings, while `separator` is a keyword-only argument because it appears after `*words`.

## 4.3 Pass multiple lists

````python
def combine_lists(*lists):
    combined = []
    for current_list in lists:
        combined.extend(current_list)
    return combined


list_1 = [1, 2]
list_2 = [3, 4]
list_3 = [5, 6]

print(combine_lists(list_1, list_2, list_3))
````

## 4.4 More complex positional arguments

A function can have required positional parameters, default parameters, and `*args`.

````python
def create_report(title, *sections, uppercase=False):
    report_title = title.upper() if uppercase else title
    body = "\n".join(f"{index}. {section}" for index, section in enumerate(sections, start=1))
    return f"{report_title}\n{body}"


print(create_report(
    "Weekly Report",
    "Completed the Python lesson",
    "Reviewed student exercises",
    "Prepared the database example",
    uppercase=True,
))
````

# 5. Variable-length keyword arguments: `**kwargs`

`**kwargs` collects extra keyword arguments into a dictionary.

The name `kwargs` is conventional, but the `**` is the important part.

````python
def show_options(**kwargs):
    print("Type:", type(kwargs))
    print("Options:", kwargs)


show_options(theme="dark", language="Python", autosave=True)
````

## 5.1 Retrieve keyword-based arguments

````python
def analyse_options(**options):
    for key, value in options.items():
        print(f"{key}: {value}")

    language = options.get("language", "Not supplied")
    difficulty = options.get("difficulty", "beginner")

    return {
        "language": language,
        "difficulty": difficulty,
        "number_of_options": len(options),
    }


summary = analyse_options(
    language="Python",
    difficulty="intermediate",
    duration_minutes=90,
    practical=True,
)

print("\nSummary:", summary)
````

Using `.get()` allows us to retrieve a value and provide a fallback when the key is missing.

## 5.2 Combine required parameters, defaults, `*args`, and `**kwargs`

````python
def create_course(
    title,
    *topics,
    instructor="Not assigned",
    duration_hours=1,
    **metadata,
):
    return {
        "title": title,
        "topics": list(topics),
        "instructor": instructor,
        "duration_hours": duration_hours,
        "metadata": metadata,
    }


course = create_course(
    "Python Functions",
    "positional arguments",
    "keyword arguments",
    "*args",
    "**kwargs",
    instructor="Dr Lee",
    duration_hours=2,
    level="introductory",
    delivery="in person",
)

course
````

# 6. The star and double-star operators

The same symbols are also used to unpack existing collections when calling functions.

- `*` unpacks a list or tuple into positional arguments.
- `**` unpacks a dictionary into keyword arguments.

````python
def calculate_total(price, quantity, discount=0):
    subtotal = price * quantity
    return subtotal - discount


positional_values = [25, 4]
keyword_values = {"discount": 10}

print(calculate_total(*positional_values, **keyword_values))
````

## 6.1 Unpack several collections

````python
numbers_1 = [1, 2, 3]
numbers_2 = [4, 5, 6]

print(calculate_sum(*numbers_1, *numbers_2))
````

## 6.2 Merge dictionaries with `**`

````python
basic_settings = {
    "language": "Python",
    "difficulty": "beginner",
}

extra_settings = {
    "duration_minutes": 60,
    "practical": True,
}

all_settings = {
    **basic_settings,
    **extra_settings,
}

print(all_settings)
````

When dictionaries contain the same key, the value appearing later wins.

````python
first = {"theme": "light", "font_size": 12}
second = {"theme": "dark"}

merged = {**first, **second}
print(merged)
````

# 7. Query a database using multiple filters and options

This example uses Python's built-in `sqlite3` module, so no installation is required.

We will:

1. create a small in-memory database,
2. insert sample students,
3. build a safe query from optional filters,
4. add sorting and limiting options.

````python
import sqlite3

connection = sqlite3.connect(":memory:")
connection.row_factory = sqlite3.Row

connection.execute(
    '''
    CREATE TABLE students (
        student_id INTEGER PRIMARY KEY,
        name TEXT NOT NULL,
        course TEXT NOT NULL,
        grade REAL NOT NULL,
        active INTEGER NOT NULL
    )
    '''
)

sample_students = [
    (1, "Aisha", "Python", 88.0, 1),
    (2, "Ben", "Databases", 74.5, 1),
    (3, "Chen", "Python", 91.0, 1),
    (4, "Dina", "Web Development", 67.0, 0),
    (5, "Ethan", "Databases", 82.0, 1),
    (6, "Fatima", "Python", 79.5, 0),
]

connection.executemany(
    '''
    INSERT INTO students (student_id, name, course, grade, active)
    VALUES (?, ?, ?, ?, ?)
    ''',
    sample_students,
)

connection.commit()
print("Database created.")
````

The function below supports several optional filters:

- `course`
- `active`
- `min_grade`
- `max_grade`
- `name_contains`

It also supports the options:

- `sort_by`
- `descending`
- `limit`

Values are passed separately from the SQL text, which avoids inserting user values directly into the query.

````python
def query_students(
    connection,
    *,
    course=None,
    active=None,
    min_grade=None,
    max_grade=None,
    name_contains=None,
    sort_by="student_id",
    descending=False,
    limit=None,
):
    allowed_sort_columns = {
        "student_id",
        "name",
        "course",
        "grade",
        "active",
    }

    if sort_by not in allowed_sort_columns:
        raise ValueError(f"sort_by must be one of: {sorted(allowed_sort_columns)}")

    conditions = []
    parameters = []

    if course is not None:
        conditions.append("course = ?")
        parameters.append(course)

    if active is not None:
        conditions.append("active = ?")
        parameters.append(int(active))

    if min_grade is not None:
        conditions.append("grade >= ?")
        parameters.append(min_grade)

    if max_grade is not None:
        conditions.append("grade <= ?")
        parameters.append(max_grade)

    if name_contains is not None:
        conditions.append("name LIKE ?")
        parameters.append(f"%{name_contains}%")

    sql = "SELECT student_id, name, course, grade, active FROM students"

    if conditions:
        sql += " WHERE " + " AND ".join(conditions)

    direction = "DESC" if descending else "ASC"
    sql += f" ORDER BY {sort_by} {direction}"

    if limit is not None:
        if not isinstance(limit, int) or limit <= 0:
            raise ValueError("limit must be a positive integer")
        sql += " LIMIT ?"
        parameters.append(limit)

    rows = connection.execute(sql, parameters).fetchall()
    return [dict(row) for row in rows]
````

## 7.1 Query using one filter

````python
query_students(connection, course="Python")
````

## 7.2 Query using multiple filters

````python
query_students(
    connection,
    course="Python",
    active=True,
    min_grade=80,
)
````

## 7.3 Query using filters and options

````python
query_students(
    connection,
    min_grade=70,
    name_contains="a",
    sort_by="grade",
    descending=True,
    limit=3,
)
````

## 7.4 Pass database filters using `**`

A dictionary can store filters and options. The dictionary is then unpacked into named arguments.

````python
student_filters = {
    "course": "Databases",
    "active": True,
    "min_grade": 70,
    "sort_by": "grade",
    "descending": True,
}

query_students(connection, **student_filters)
````

# 8. Creating tasks

We will use a `Task` class and a `create_task()` helper function.

Each task can contain multiple notes.

````python
from dataclasses import dataclass, field


@dataclass
class Task:
    title: str
    completed: bool = False
    notes: list[str] = field(default_factory=list)

    def add_note(self, note):
        self.notes.append(note)

    def mark_complete(self):
        self.completed = True
````

## 8.1 Python gotcha: mutable default arguments

The following function is intentionally incorrect:

```python
def create_task_bad(title, notes=[]):
    ...
```

The list used as the default value is created once when Python defines the function. It is not recreated for every call. Therefore, separate calls can unexpectedly share the same list.

````python
def create_task_bad(title, notes=[]):
    return Task(title=title, notes=notes)


bad_task_1 = create_task_bad("Prepare lesson")
bad_task_1.add_note("Add examples for positional arguments")
bad_task_1.add_note("Add an example for keyword arguments")

bad_task_2 = create_task_bad("Review exercises")

print("Task 1 notes:", bad_task_1.notes)
print("Task 2 notes:", bad_task_2.notes)
print("Same list object:", bad_task_1.notes is bad_task_2.notes)
````

### What happened?

`bad_task_1` and `bad_task_2` received the same precomputed default list.

The default parameter value was evaluated once when the function was defined. Adding notes through one task changed the list seen by the other task.

This problem can occur with mutable defaults such as:

- lists
- dictionaries
- sets
- class instances

Use `None` as the default, then create a new object inside the function.

## 8.2 Correct solution: use `None`

````python
def create_task(title, notes=None, completed=False):
    if notes is None:
        notes = []
    else:
        # Copy the supplied list so the task owns its own notes.
        notes = list(notes)

    return Task(
        title=title,
        completed=completed,
        notes=notes,
    )
````

## 8.3 Create Task 1, add multiple notes, and create Task 2

````python
task_1 = create_task("Prepare Python notebook")
task_1.add_note("Explain positional, keyword, and default arguments")
task_1.add_note("Demonstrate *args and **kwargs")
task_1.add_note("Include the mutable default argument gotcha")

task_2 = create_task("Upload notebook to Colab")
task_2.add_note("Open the notebook in Google Colab")
task_2.add_note("Run all cells from top to bottom")

print(task_1)
print(task_2)
print("Separate note lists:", task_1.notes is not task_2.notes)
````

## 8.4 Create a task with initial notes

````python
task_3 = create_task(
    "Teach database filtering",
    notes=[
        "Create a sample SQLite database",
        "Use parameterised SQL values",
    ],
)

task_3.add_note("Demonstrate several filters at the same time")
print(task_3)
````

# 9. Creating students

The following function combines:

- required positional arguments,
- default arguments,
- variable keyword arguments.

````python
def create_student(
    student_id,
    name,
    course="Python",
    active=True,
    **additional_details,
):
    student = {
        "student_id": student_id,
        "name": name,
        "course": course,
        "active": active,
    }

    student.update(additional_details)
    return student


student_1 = create_student(
    101,
    "Maya",
    grade=86,
    email="maya@example.edu",
)

student_2 = create_student(
    student_id=102,
    name="Leo",
    course="Databases",
    active=False,
    grade=73,
    campus="Brisbane",
)

print(student_1)
print(student_2)
````

````python
def build_student_query_solution(**filters):
    if not filters:
        return "Find all students."

    descriptions = [
        f"{key}={value!r}"
        for key, value in filters.items()
    ]

    return "Find students where " + ", ".join(descriptions) + "."


print(build_student_query_solution(
    course="Python",
    active=True,
    min_grade=75,
))
````
