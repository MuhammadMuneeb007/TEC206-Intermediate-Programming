```python
# ============================================================
# 1. DIFFERENT DATA TYPES IN PYTHON
# ============================================================

# A data type describes the kind of value stored in a variable.

# Integer (int): whole numbers
age = 28
number_of_students = 30

# Float (float): decimal numbers
height = 1.75
temperature = 24.5

# String (str): text
name = "Muhammad"
course = "Python Programming"

# Boolean (bool): True or False
is_student = True
has_graduated = False

# List (list): ordered and changeable collection
subjects = ["Python", "Databases", "Networking"]

# Tuple (tuple): ordered and normally unchangeable collection
coordinates = (10, 20)

# Dictionary (dict): key-value pairs
student_information = {
    "name": "Ali",
    "age": 22,
    "course": "Computer Science"
}

# Set (set): collection of unique values
student_ids = {101, 102, 103, 101}

# NoneType: represents the absence of a value
result = None


print("Integer:", age)
print("Float:", height)
print("String:", name)
print("Boolean:", is_student)
print("List:", subjects)
print("Tuple:", coordinates)
print("Dictionary:", student_information)
print("Set:", student_ids)
print("None:", result)
```

```python
# ============================================================
# 2. CHECKING DATA TYPES
# ============================================================

age = 28
height = 1.75
name = "Muhammad"
is_student = True
subjects = ["Python", "Databases"]
coordinates = (10, 20)
student = {"name": "Ali", "age": 22}
student_ids = {101, 102, 103}
result = None


# type() displays the exact data type
print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
print(type(subjects))
print(type(coordinates))
print(type(student))
print(type(student_ids))
print(type(result))


# isinstance() checks whether a value belongs to a particular type
print(isinstance(age, int))
print(isinstance(height, float))
print(isinstance(name, str))
print(isinstance(subjects, list))


# Checking a type inside an if statement
if isinstance(age, int):
    print("age contains an integer")

if isinstance(name, str):
    print("name contains a string")


# Check whether a value is either an integer or a float
score = 89.5

if isinstance(score, (int, float)):
    print("score is a numeric value")
else:
    print("score is not a numeric value")
```

```python
# ============================================================
# 3. IMPORTANCE OF DATA TYPES
# ============================================================

# Data types determine which operations can be performed.

# Integer addition
number_one = 10
number_two = 20

total = number_one + number_two
print(total)  # 30


# String addition joins text
first_name = "Muhammad"
last_name = "Muneeb"

full_name = first_name + " " + last_name
print(full_name)


# A string and an integer cannot be added directly
age = 28

# Incorrect:
# message = "My age is " + age

# Correct: convert the integer to a string
message = "My age is " + str(age)
print(message)


# Numeric values stored as strings must be converted before calculation
first_value = "10"
second_value = "20"

incorrect_result = first_value + second_value
print(incorrect_result)  # 1020

correct_result = int(first_value) + int(second_value)
print(correct_result)  # 30


# A list can contain different types
mixed_data = ["Ali", 22, 85.5, True]
print(mixed_data)


# However, using consistent types often makes data easier to process
student_names = ["Ali", "Sara", "John"]
student_ages = [22, 20, 24]

print(student_names)
print(student_ages)


# Python variables can be reassigned to another type
value = 100
print(value, type(value))

value = "One hundred"
print(value, type(value))
```

```python
# ============================================================
# 5. CLASS WITH DATA AND BEHAVIOURS
# ============================================================

# A class can contain:
#
# 1. Data, also called attributes
# 2. Behaviours, also called methods


class Car:

    def __init__(self, brand, colour):
        # Data or attributes
        self.brand = brand
        self.colour = colour

    def move(self):
        # Behaviour or method
        print(self.brand, "is moving")

    def stop(self):
        # Behaviour or method
        print(self.brand, "has stopped")


# Instantiate an object
my_car = Car("Toyota", "White")


# Access the object's data
print("Brand:", my_car.brand)
print("Colour:", my_car.colour)


# Use the object's behaviours
my_car.move()
my_car.stop()
```

```python
# ============================================================
# 6. PERSON CLASS
# ============================================================


class Person:

    def __init__(self, name, age, occupation):
        # Person data
        self.name = name
        self.age = age
        self.occupation = occupation

    def introduce(self):
        # Person behaviour
        print(
            "Hello, my name is",
            self.name,
            "and I work as a",
            self.occupation
        )

    def celebrate_birthday(self):
        # Change the object's age
        self.age = self.age + 1
        print(self.name, "is now", self.age, "years old")


# Instantiate a Person object
person_one = Person(
    name="Muhammad",
    age=28,
    occupation="Researcher"
)


print("Name:", person_one.name)
print("Age:", person_one.age)
print("Occupation:", person_one.occupation)

person_one.introduce()
person_one.celebrate_birthday()
```

```python
# ============================================================
# 7. STUDENT CLASS: STRUCTURE, DATA, AND BEHAVIOURS
# ============================================================

# The class defines the structure of a Student object.
#
# Data:
# - name
# - student_id
# - course
# - marks
#
# Behaviours:
# - study
# - display_information
# - add_mark
# - calculate_average
# - check_result


class Student:

    def __init__(self, name, student_id, course):
        # Data or attributes
        self.name = name
        self.student_id = student_id
        self.course = course
        self.marks = []

    def study(self):
        # Behaviour
        print(self.name, "is studying", self.course)

    def display_information(self):
        # Behaviour
        print("Name:", self.name)
        print("Student ID:", self.student_id)
        print("Course:", self.course)
        print("Marks:", self.marks)

    def add_mark(self, mark):
        # Behaviour that updates the object's data
        if 0 <= mark <= 100:
            self.marks.append(mark)
            print(mark, "was added")
        else:
            print("The mark must be between 0 and 100")

    def calculate_average(self):
        # Behaviour that performs a calculation
        if len(self.marks) == 0:
            return 0

        return sum(self.marks) / len(self.marks)

    def check_result(self):
        # Behaviour that makes a decision
        average = self.calculate_average()

        if average >= 50:
            return "Pass"

        return "Fail"


# Instantiate a Student object
student_one = Student(
    name="Ali",
    student_id="S101",
    course="Computer Science"
)


student_one.add_mark(80)
student_one.add_mark(75)
student_one.add_mark(90)

student_one.display_information()
student_one.study()

print("Average:", student_one.calculate_average())
print("Result:", student_one.check_result())
```

```python
from graphviz import Digraph

# Create a simple vertical flowchart
chart = Digraph("Classes")
chart.attr(rankdir="TB")
chart.attr("node", shape="box", style="rounded", fontsize="12")

# Main idea
chart.node("A", "Class")
chart.node("B", "Blueprint for creating objects")

chart.edge("A", "B")

# What a class contains
chart.node("C", "Data\nAttributes")
chart.node("D", "Behaviours\nMethods")

chart.edge("B", "C")
chart.edge("B", "D")

# Simple examples
chart.node("C1", "name")
chart.node("C2", "age")

chart.edge("C", "C1")
chart.edge("C", "C2")

chart.node("D1", "introduce()")
chart.node("D2", "study()")

chart.edge("D", "D1")
chart.edge("D", "D2")

# Objects created from the class
chart.node("E", "Create Objects")

chart.edge("B", "E")

chart.node("F", "Person Object 1\nName: Ali\nAge: 22")
chart.node("G", "Person Object 2\nName: Sara\nAge: 25")

chart.edge("E", "F")
chart.edge("E", "G")

# Final explanation
chart.node(
    "H",
    "Each object has its own data,\nbut uses the same class methods"
)

chart.edge("F", "H")
chart.edge("G", "H")

# Display the flowchart
chart
```
